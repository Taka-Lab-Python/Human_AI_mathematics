【問題5】ベクトルの半時計まわりの回転

```python
import numpy as np
import matplotlib.pyplot as plt

w = np.array([[1, 1],
              [4, 3]])
print("ベクトルの初期の座標点","\n",  w)

# 2. 1次変換のために転置（.T）して、数式・グラフ用の形にする
w_T = w.T

# 3. 回転行列 R の作成（反時計回りに45度）
theta = np.radians(45)
R = np.array([
    [ np.    , -np.   ],
    [ np.    ,  np.   ]
])

# 4. 1次変換を実行（行列の積）
w_rotated = R @ w_T

# --- グラフ描画 ---
plt.figure(figsize=(6, 6))

# 元の直線（青色）
# w_T[0] は [1, 7] (X座標)、w_T[1] は [1, 1] (Y座標)
plt.plot(w_T[0], w_T[1], marker='o', color='blue', label='Original Line (a-b)')

# 回転後の直線（赤色・点線）
# w_rotated[0] が回転後のX座標、w_rotated[1] が回転後のY座標
plt.plot(w_rotated[0], w_rotated[1], marker='o', color='red', linestyle='--')

# 原点(0,0)のプロット
plt.plot(0, 0, marker='x', color='black', markersize=10, label='Origin')

font_size = 14
plt.xticks(fontsize=font_size)
plt.xlabel("X", fontsize=font_size)
plt.ylabel("Y", fontsize=font_size)
plt.yticks(fontsize=font_size)
plt.xlim(-1, 5)
plt.ylim(0, 6)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 四角の４点
a = [1, 1]
b = [7, 1]
c = [7, 2]
d = [1, 2]

# (2行5列)
points = np.array([a, b, c, d, a]).T
print(points)

# 2. 回転行列 R の作成 (例として反時計回りに45度回転させます)
theta = np.radians(45)  # 度数法からラジアンに変換
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

# 3. 1次変換を実行（行列の掛け算： R × 特徴点）
transformed_points = R @ points

# --- グラフ描画 ---
plt.figure(figsize=(8, 6))

# 元の長方形をプロット（青色）
plt.plot(points[0, :], points[1, :],
         color='blue', marker='o', linewidth=2, label='Original Rectangle')
# 頂点にラベルを表示 (a, b, c, d)
labels = ['a', 'b', 'c', 'd']
for i, label in enumerate(labels):
    plt.text(points[0, i]+0.1, points[1, i]+0.1, label, color='blue', fontsize=12)

# 1次変換（回転）後の長方形をプロット（赤色）
plt.plot(transformed_points[0, :], transformed_points[1, :],
         color='red', marker='o', linewidth=2, linestyle='--', label='Rotated Rectangle (45°)')
# 回転後の頂点にラベルを表示 (a', b', c', d')
for i, label in enumerate(labels):
    plt.text(transformed_points[0, i]+0.1, transformed_points[1, i]+0.1, f"{label}'", color='red', fontsize=12)

# 原点のプロット
plt.plot(0, 0, marker='x', color='black', markersize=10, label='Origin (0,0)')

# グラフの見た目調整
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(0, color='black', linewidth=0.8, linestyle=':')
plt.xlim(-5, 9)
plt.ylim(-2, 8)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11, loc='upper left')
plt.title('Linear Transformation: Rotating a Rectangle', fontsize=14)

# 幾何学的に正しい回転に見せるため、縦横比を1:1に固定
plt.gca().set_aspect('equal', adjustable='box')

plt.show()
```
