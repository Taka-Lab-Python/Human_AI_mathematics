```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# --- 1. テスト用のRGB画像(70x70)をPillowで作る ---
# 学生がすぐ実行できるように、プログラム内でグラデーション画像を生成します
grid = np.linspace(0, 255, 70).astype(np.uint8)
x, y = np.meshgrid(grid, grid)
rgb_array = np.stack([x, y, 255 - x], axis=-1)

# NumPyの数字の塊を、Pillowの「画像オブジェクト」に変換
img_rgb = Image.fromarray(rgb_array)

# --- 2. Pillowを使って画像を加工する ---
# 【回転】時計回りに90度回転（Pillowなら直感的に書けます）
# ※Pillowのrotateは反時計回りが基本なので、-90を指定すると時計回りになります
img_rotated = img_rgb.rotate(-90)

# 【モノクロ変換】RGBからグレースケール（L: Luminance）に変換
#img_gray = img_rotated.convert('L')

# --- 3. 画像を「70×70の数値データ（行列）」に書き出す ---
# Pillowの画像をNumPyの配列に戻すことで、中身の数字を取り出します
data_matrix = np.array(img_rotated)

# 試しに、コンソールに70x70の行列データを書き出してみる
print("--- 70x70のモノクロ数値データ（行列） ---")
print("データの形状 (縦, 横):", data_matrix.shape)
print(data_matrix)

# --- 4. Matplotlibを使って結果を画面に表示する ---
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# 左：元のカラー画像
axes[0].imshow(img_rgb)
axes[0].set_title("Original RGB Image")
axes[0].grid(False)

# 右：回転してモノクロにした画像（cmap='gray' で白黒表示にします）
axes[1].imshow(img_rotated)
axes[1].set_title("Rotated & Monochrome Image")
axes[1].grid(False)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageChops  # 反転用に ImageChops を追加します

# 1. 画像の読み込み (Matplotlibを使用)
path = "/content/drive/MyDrive/11_IT-college/AI数学/線形代数学/画像/s_2.jpg"
img_array = plt.imread(path)

# 2. NumPyの配列からPillowの「画像オブジェクト」に変換
img_pil = Image.fromarray(img_array)

# 【追加】50x40のモノクロ（グレースケール）画像に変換します
img_gray = img_pil.resize((40, 50)).convert('L')

# 3. Pillowの機能で、白と黒をガシャリと反転させる
#（すべてのセルの数字を「255 - 現在の値」にする処理を自動で行ってくれます）
img_inverted = ImageChops.invert(img_gray)

# 4. 反転した画像を「50×40の数値データ（行列）」に書き出す
data_matrix = np.array(img_inverted)

# NumPyの省略をオフにして、50行40列の全数字を出力する設定
np.set_printoptions(threshold=np.inf)
np.set_printoptions(linewidth=200)

print("--- 白黒反転後の 50x40 の数値データ（行列） ---")
print("データの形状 (縦, 横):", data_matrix.shape)
print(data_matrix)

# --- 5. 画面に確認用の画像を表示する ---
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# 左：元のモノクロ画像
axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title("Original Monochrome (50x40)")
axes[0].grid(False)

# 右：白黒を反転した画像
axes[1].imshow(img_inverted, cmap='gray')
axes[1].set_title("Inverted Image (White <-> Black)")
axes[1].grid(False)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# 1. 画像の読み込みと50x40モノクロへの前処理
path = "/content/drive/MyDrive/11_IT-college/AI数学/線形代数学/画像/s_2.jpg"
img_array = plt.imread(path)
img_pil = Image.fromarray(img_array).resize((40, 50)).convert('L')
src_matrix = np.array(img_pil)  # 元の50x40の行列

# 2. 反転後の数字を入れるための「空の50x40の行列」をはじめに用意する
#（最初はすべて0が入った真っ黒な台紙を作っておきます）
dst_matrix = np.zeros((50, 40))

# 3. 【核心】2重のfor文で、1マスずつ数字をひっくり返す！
# 縦（行）のループ：0行目から49行目まで
for y in range(50):
    # 横（列）のループ：0列目から39列目まで
    for x in range(40):
        # 元の場所の数字（明るさ）を読み取る
        original_pixel = src_matrix[y, x]

        # 255から引き算して、白黒を反転させた数字を作る
        #（0なら255に、255なら0に変換）
        inverted_pixel = 255 - original_pixel

        # 新しい行列の同じ場所に、反転させた数字を書き込む
        dst_matrix[y, x] = inverted_pixel

# 4. 全数字をコンソールに書き出す設定
np.set_printoptions(threshold=np.inf)
np.set_printoptions(linewidth=200)

print("--- 2重のfor文で反転させた 50x40 の数値データ ---")
print(dst_matrix)

# --- 5. 画面に表示して確認 ---
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(src_matrix, cmap='gray')
axes[0].set_title("Original (50x40)")
axes[0].grid(False)

axes[1].imshow(dst_matrix, cmap='gray')
axes[1].set_title("Inverted by 2D for-loop")
axes[1].grid(False)

plt.show()
```
