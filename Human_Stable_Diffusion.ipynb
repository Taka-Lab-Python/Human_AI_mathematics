```python
# 1.1 画像の生成の前の準備
# GPU が使えるかを判定（Colab なら cuda）
# stable diffusion1.5

import torch
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("*** cuda using ***")
else:
    device = torch.device("cpu")
    print("*** CPU using ***")
```

```python
"""
１　<<<"stable-diffusion-v1-5/stable-diffusion-v1-5">>>
テキスト入力 ⇒ トークナイズ（文章 → トークン） ⇒　テキストエンコーダ（CLIP など）
⇒　ノイズからの逐次的生成（拡散過程）　⇒　UNet によるノイズ予測
VAE による潜在表現 ⇒　画像への復元
"""

import torch
from diffusers import StableDiffusionPipeline

pipe = StableDiffusionPipeline.from_pretrained(
    "stable-diffusion-v1-5/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
    variant="fp16",).to(device)
```

```python
# ２　<<<絵画芸術　Lykon/DreamShaper)>>>

import torch
from diffusers import StableDiffusionPipeline

pipe = StableDiffusionPipeline.from_pretrained(
    "Lykon/DreamShaper",
    torch_dtype=torch.float16,).to(device)
```

```python
"""
prompt (str)
negative_prompt (str)：出したくない、表現したくない要素
num_inference_steps (int)： 大きいほど繊細　20～50
guidance_scale　(float) ：忠実度　7.0～12.0
width / height　(int)：　512, 768
num_images_per_prompt=1  1枚だけ作成させる
"""

prompt = "a photgraph of a mouse on the desk"
n_prompt = "Black and white photo"
pipe(prompt, num_images_per_prompt=1, num_inference_steps=20,
     negative_prompt=n_prompt, guidance_scale=7).images[0]


```

```python
# 日本語プロンプト

pipe = StableDiffusionPipeline.from_pretrained(
    "rinna/japanese-stable-diffusion",
    torch_dtype=torch.float16).to(device)

prompt_jp = "馬に乗った宇宙飛行士の写真"
pipe(prompt_jp).images[0]
```
