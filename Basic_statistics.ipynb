```python
import numpy as np
weight = np.array([49.5, 46.6, 56.9, 47.8, 51.2, 63.8, 50.8, 40.6,
                   43.4, 52.6, 56.4, 52.1, 57.4, 47.5, 64.7])

height = np.array([143, 149.7, 158.6, 141.9, 160.5, 176.6, 155.7, 144.9,
                   133.8, 157.3, 163.1, 157.9, 176.9, 150.3, 177.4])
```

```python
def average_calc(value):
  n = len(value)
  mean_val = np.mean(value)
  max_val = np.max(value)
  min_val = np.min(value)
  var_val = np.var(value)
  std_val = np.std(value)

  print("データ数=", n)
  print(f"平均値={mean_val:.1f}")
  print(f"分散={var_val:.1f}")
  print(f"標準偏差={std_val:.1f}")
  print(f"最大値={max_val:.1f}")
  print(f"最小値={mean_val:.1f}")
  return

average_calc(height)
```

```python
# 箱ひげ図
import seaborn as sns

sns.boxplot(data=height, color="blue")
plt.xlabel("身長の箱ひげ図", fontsize=16)
plt.ylabel("身長　(cm)", fontsize=16)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.show()
```

```python
from matplotlib import pyplot as plt
from scipy.stats import binom

plt.figure(figsize=(4, 3))

# パラメータの設定
n = 10   # 試行回数
p = 0.5  # 成功確率

x = np.arange(0, n+1)  # 二項分布の確率質量関数の計算
pmf = binom.pmf(x, n, p)
plt.bar(x, pmf, align='center', alpha=0.7, color='blue')
plt.title('Binomial Distribution PMF')
plt.xlabel('成 功 確 率')
plt.ylabel('確　率')
plt.grid(True)
plt.show()
```

正規分布表を計算する

```python
from scipy import stats

# 1. Z値から確率を求める（PDFの表の代わり）
Z= 1.0
prob = stats.norm.cdf(Z)
print(f"確率: {prob:.4f}")
```

```python
# 正規分布表の計算プログラム
from scipy import stats

# 2. 確率からZ値を逆引きする
table = 0.8413
z_value = stats.norm.ppf(table)
print(f"Z値: {z_value:.3f}")
```
